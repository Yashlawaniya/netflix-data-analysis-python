## Objective
To analyze the distribution of Movies and TV Shows
in the Netflix dataset and visualize the results.


### Import the libraries

```python
import pandas as pd
import matplotlib.pyplot as plt

```

### Load the cleaned Netflix dataset

```python
df = pd.read_csv("../../data/processed/netflix_cleaned.csv")

df.head()
```

### Verify that the data loaded correctly

```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nContent types:")
print(df["type"].value_counts())
```

### Calculate content percentages

```python
#Count each content type
content_counts = df["type"].value_counts()

#Calculate percentage of each content type
content_percentage = (content_counts/len(df)) * 100

#Display the results
print("Content Counts: \n",content_counts)

print("\nContent Percentages: \n",content_percentage.round(2))
```

### Creating a bar chart using Matplotlib

```python
plt.Figure(figsize=(8,5))

plt.bar(content_counts.index , content_counts.values)
plt.bar_label(plt.gca().containers[0])

plt.title("Distribution of Netflix Content Types")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.tight_layout()
plt.savefig("../../reports/figures/netflix_content_type_bar.png",
            dpi=300, bbox_inches="tight")
plt.show()
```

### Create a pie chart 

```python
plt.Figure(figsize=(7,7))

plt.pie(
    content_counts.values,
    labels=content_counts.index,
    autopct="%1.2f%%",
    startangle=90
)

plt.title("Netflix Content Type Distribution")
plt.tight_layout()
plt.savefig("../../reports/figures/netflix_content_type_pie.png",
            dpi=300, bbox_inches="tight")
plt.show()
```
