## Step 3: Missing Value Analysis

Missing values are records where one or more columns do not contain a value. These can affect the accuracy of analysis and reporting.

In this step, we calculate the number and percentage of missing values in each column. We will use the results to decide how to handle missing data without unnecessarily losing useful information.

```python
# Check missing values in each column

#Count missing values in each column. 
missing_values = df.isnull().sum()

# Calculate missing value percentages. 
missing_percentage = (missing_values/len(df))*100

# Create a summary table with counts and percentages. 
missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

# Sort columns by missing value count, highest first. 
missing_summary = missing_summary.sort_values(
    by="Missing Values",
    ascending=False
)

missing_summary
```

## Step 5: Date Column Analysis

### Step 5.1: Inspecting the Date Column

The `date_added` column currently has a string data type. Before converting it, we will inspect sample values to understand their format and identify any unusual entries.

### Step 5.2: Converting the Date Column

The `date_added` column currently stores dates as strings. We will convert it into Pandas datetime format using `pd.to_datetime()`.

This conversion will allow us to extract the year and month, sort dates chronologically, and perform time-based analysis more reliably.

```python

# Step 5.2: Convert the date column

# Load the original dataset again
df = pd.read_csv("../../data/raw/Dataset.csv")

# Convert date_added into datetime format
df["date_added"] = pd.to_datetime(
    df["date_added"],
    format="%m/%d/%Y",
    errors="coerce"
)

# Display the first 10 converted dates
print("First 10 converted dates:")
print(df["date_added"].head(10))

# Check the column data type
print("\nData type:")
print(df["date_added"].dtype)

# Count dates that could not be converted
print("\nInvalid or missing dates:")
print(df["date_added"].isna().sum())
```

## Step 6: Duration Column Analysis

```python
#Display the first 10 duration values. 
print("First 10 duration values:")
print(df["duration"].head(10))

# Display the different duration units. 
print("\nDuration unit:")
print(df["duration"].str.extract(r"([A-Za-z]+)$")[0].value_counts())
```

### Step 6.1: Standardizing Duration Information

The `duration` column contains numeric values combined with units. Movies generally use minutes, whereas TV Shows use seasons.

We will extract the numeric value and its unit into separate columns. This will make duration information easier to analyze while preserving the original column. The singular and plural season labels will also be standardized.

```python
df["duration_value"] = pd.to_numeric(
    df["duration"].str.extract(r"(\d+)")[0],
    errors="coerce"
).astype("Int64")

df["duration_unit"] = df["duration"].str.extract(r"([A-Za-z]+)$")[0]

df["duration_unit"] = df["duration_unit"].replace(
    {"Seasons" : "Season"}
)

df[["type", "duration", "duration_value", "duration_unit"]].head(10)
```
